```python
import pandas as pd
import yfinance as yf
from datetime import datetime, timedelta

print("Downloading fresh stock data...")

stocks = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'META', 'TSLA', 'JPM', 'BAC', 'WMT', 'PFE']
end_date = datetime.now()
start_date = end_date - timedelta(days=365)

all_data = []

for symbol in stocks:
    df = yf.download(symbol, start=start_date, end=end_date, progress=False)
    df = df.reset_index()  # Move Date from index to column
    df.columns = [col[0] if isinstance(col, tuple) else col for col in df.columns]
    df['Symbol'] = symbol
    all_data.append(df)
    print(f"  {symbol}: {len(df)} days")

stock_df = pd.concat(all_data, ignore_index=True)

# Clean column names
stock_df.columns = stock_df.columns.str.strip()
print(f"\nColumns: {stock_df.columns.tolist()}")
print(f"Shape: {stock_df.shape}")
print(stock_df.head(3))

# Save clean version
stock_df.to_csv('data/raw/stock_prices_clean.csv', index=False)
print("\n✅ Saved clean data!")
```

```python
import pandas as pd
import numpy as np
import random

random.seed(42)
np.random.seed(42)

stocks = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'META', 'TSLA', 'JPM', 'BAC', 'WMT', 'PFE']

templates = [
    "{stock} Hits All-Time High on Strong Earnings",
    "{stock} Shares Drop After Weak Guidance",
    "{stock} Announces New Product Line",
    "Analysts Upgrade {stock} to Buy",
    "{stock} Reports Record Quarterly Revenue",
    "{stock} CFO Steps Down",
    "{stock} Expands into New Markets",
    "FDA Approves {stock} Drug Application",
    "{stock} Beats Earnings Estimates",
    "Market Rally Lifts {stock} Higher",
    "{stock} Announces Stock Buyback",
    "{stock} Misses Q3 Targets",
    "Tech Sell-off Drags {stock} Lower",
    "{stock} Partners with Industry Leader",
    "Insider Buying at {stock}",
]

publishers = ['Reuters', 'Bloomberg', 'CNBC', 'Wall Street Journal', 
              'MarketWatch', 'Financial Times', 'InvestorPlace', 'Zacks']

# Create 2000 news articles across 2025
start = '2025-01-01'
end = '2025-12-31'
dates = pd.date_range(start, end, freq='D')

news_data = []
for _ in range(2000):
    stock = random.choice(stocks)
    headline = random.choice(templates).format(stock=stock)
    pub_date = random.choice(dates)
    hour = random.randint(7, 19)
    minute = random.randint(0, 59)
    timestamp = f"{pub_date.date()} {hour:02d}:{minute:02d}:00"
    
    news_data.append({
        'headline': headline,
        'stock': stock,
        'date': timestamp,
        'publisher': random.choice(publishers),
        'url': f'https://finance.example.com/news/{len(news_data)}'
    })

news_df = pd.DataFrame(news_data)
news_df['date'] = pd.to_datetime(news_df['date'])
news_df = news_df.sort_values('date').reset_index(drop=True)

news_df.to_csv('data/raw/financial_news.csv', index=False)

print(f"✅ Created {len(news_df)} news articles")
print(f"Date range: {news_df['date'].min()} to {news_df['date'].max()}")
print(f"\nSample headlines:")
print(news_df['headline'].sample(5).tolist())
```
